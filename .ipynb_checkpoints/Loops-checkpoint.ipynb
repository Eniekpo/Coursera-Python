```python
def attempts(n):
    x = 1
    while x <= n:
        print("Attempt " + str(x))
        x += 1
    print("Done")

attempts(5)
```

```python
def get_username():
    return input("Enter a username: ")


def valid_username(username):
    return username.isalnum() and len(username) >= 3


username = get_username()
while not valid_username(username):
    print("Invalid username")
    username = get_username()

#This code will give an error because get_username is not defined
```

```python
x = 1
sum = 0
while x < 10:
    sum = sum + x
    x = x + 1

product = 1
while x < 10:
    product = product * x
    x = x + 1

print(sum, product)
#Ouput 45 1
```
